```python
CATALOG = "fk_ce_2026_07_1382"

# H2 – Run protocol: checkpoint_counts table
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.ops.checkpoint_counts (
  checkpoint_name STRING,
  table_name      STRING,
  row_count       BIGINT,
  captured_at     TIMESTAMP
)
""")

# H3 – Load audit: load_audit table
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {CATALOG}.ops.load_audit (
  job_run_id   STRING,
  taskname     STRING,
  target_table STRING,
  row_count    BIGINT,
  completed_at TIMESTAMP
)
""")

print("H2 checkpoint_counts and H3 load_audit tables created.")
```

```python
CATALOG = "fk_ce_2026_07_1382"

# ── Verify H2 & H3 tables exist with correct schema ──
print("=== H2: ops.checkpoint_counts ===")
display(spark.sql(f"DESCRIBE TABLE {CATALOG}.ops.checkpoint_counts"))

print("\n=== H3: ops.load_audit ===")
display(spark.sql(f"DESCRIBE TABLE {CATALOG}.ops.load_audit"))

print("\n=== Current row counts ===")
cc = spark.sql(f"SELECT COUNT(*) AS c FROM {CATALOG}.ops.checkpoint_counts").collect()[0][0]
la = spark.sql(f"SELECT COUNT(*) AS c FROM {CATALOG}.ops.load_audit").collect()[0][0]
print(f"checkpoint_counts rows: {cc}")
print(f"load_audit rows:       {la}")
print("\nTables are ready. H2/H3 logic is embedded in the bronze, silver & gold notebooks.")
print("The gold notebook's final cell implements H2 (13-table row counts) and H3 audit for quality_checks.")
print("Bronze and silver notebooks append H3 load_audit entries for every table they write.")
```
